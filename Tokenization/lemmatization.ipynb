```python
words=["eating","eats","eaten","writing","writes","programming","programs","history","finally","finalized"]
```

```python
from nltk.stem import WordNetLemmatizer
```

```python
lemmitizer=WordNetLemmatizer()
```

```python
for word in words:
    print(word +" --> " +lemmitizer.lemmatize(word,pos="v"))
```
